```python
import pymupdf
from pathlib import Path
```

```python
def extract_structured_pages(path):
    doc = pymupdf.open(path)
    pages = []
    
    current_chapter = "Introduction / Front Matter"
    current_section = ""

    for page_num in range(doc.page_count):
        raw_text = doc[page_num].get_text("text")
        cleaned_text = clean_page(raw_text)
        
        if cleaned_text:
            # Update running chapter and section state from page content
            current_chapter, current_section = detect_headings_in_page(
                cleaned_text, current_chapter, current_section
            )
            
            pages.append({
                "module": path.name,
                "chapter": current_chapter,
                "section": current_section,
                "page": page_num + 1,
                "text": cleaned_text
            })
            
    doc.close()
    return pages
```

```python
raw_dir = Path("../data/raw")
pdf_paths = sorted(raw_dir.glob("*.pdf"))

all_modules = {}

for pdf_path in pdf_paths:
    print(f"Extracting {pdf_path.name}...")
    pages = extract_structured_pages(pdf_path)
    all_modules[pdf_path.name] = pages
    print(f"  -> Extracted {len(pages)} pages.")
    
print(f"\nDone! Ingested {len(all_modules)} modules.")
```

```python
import re
import unicodedata

def clean_page(text: str) -> str:
    lines = text.split("\n")
    cleaned_lines = []
    
    # Check only the top 3 non-empty lines for header noise
    checked_lines = 0
    
    for line in lines:
        stripped = line.strip()
        
        if checked_lines < 3 and stripped:
            # 1. Author header in logic.pdf
            if "By: Teklay G." in stripped:
                continue
            
            # 2. Arabic page numbers: "Page 12" or standalone digits "12"
            if re.match(r"^(?:Page\s*)?\d+$", stripped, re.IGNORECASE):
                continue
                
            # 3. Roman numerals in physics.pdf: "i", "iv", "vii", "Page iii", etc.
            if re.match(r"^(?:Page\s*)?[ivxlcdm]+$", stripped, re.IGNORECASE):
                continue
                
            # If it reached here and has content, it's real body text
            checked_lines += 1
            
        cleaned_lines.append(line)
        
    text = "\n".join(cleaned_lines)
    
    # --- Text Normalization ---
    # 1. Fix ligatures (fi, fl) & Unicode
    text = unicodedata.normalize("NFKC", text)
    
    # 2. Fix odd Microsoft Word/PDF quote artifacts
    text = (text.replace("‗‗", '"').replace("‘‘", '"')
                .replace("‖", '"').replace("‗", "'"))
    
    # 3. Fix hyphenated word breaks: "de-\nfinition" -> "definition"
    text = re.sub(r"(\w+)-\s*\n\s*(\w+)", r"\1\2", text)
    
    # 4. Collapse inline whitespace & join single line-breaks (keep paragraph breaks)
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n\s*\n+", "\n\n", text)
    text = re.sub(r"(?<!\n)\n(?!\n)", " ", text)

    text = normalize_text(text)
    
    return text.strip()
```

```python
def normalize_text(text: str) -> str:
    # 1. Unicode & invisible character cleanup
    text = unicodedata.normalize("NFKC", text)
    text = text.replace("\xa0", " ").replace("\xad", "")
    text = re.sub(r"[\u200b\u200c\u200d\ufeff]", "", text)

    # 2. Quote and dash artifacts
    text = (text.replace("‗‗", '"').replace("‘‘", '"')
                .replace("‖", '"').replace("‗", "'")
                .replace("―", "—"))

    # 3. Bullets and TOC dot leaders
    text = re.sub(r"[\uf0d8\uf076\uf0b7•■◆►]\s*", "- ", text)
    text = re.sub(r"\.{3,}", "...", text)

    # 4. Rejoin hyphenated linebreaks: "infor-\nmation" -> "information"
    text = re.sub(r"(\w+)-\s*\n\s*(\w+)", r"\1\2", text)

    # 5. Spacing and paragraph structure
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n\s*\n+", "\n\n", text)
    text = re.sub(r"(?<!\n)\n(?!\n)", " ", text)

    return text.strip()
```

```python
# Regex patterns for Ethiopian university textbook structures
CHAPTER_RE = re.compile(
    r"^(?:CHAPTER|UNIT)\s+([A-Z0-9]+|ONE|TWO|THREE|FOUR|FIVE|SIX|SEVEN|EIGHT|NINE|TEN)(?:\s*[:\-])?(.*)$",
    re.IGNORECASE
)

SECTION_RE = re.compile(
    r"^(?:Lesson\s+\d+|(\d+\.\d+(?:\.\d+)?))\s*[:\.\-]?\s*(.*)$",
    re.IGNORECASE
)

def detect_headings_in_page(text: str, current_chapter: str, current_section: str):
    """
    Scans the page lines to update the active chapter and section,
    ignoring Table of Contents lines (which contain trailing '... 12').
    """
    lines = text.split("\n")
    
    for line in lines:
        stripped = line.strip()
        
        # Skip Table of Contents lines with dot leaders or page trailers
        if "..." in stripped:
            continue
            
        # 1. Check for Chapter / Unit
        chap_match = CHAPTER_RE.match(stripped)
        if chap_match and len(stripped) < 80:  # avoid long regular sentences
            current_chapter = stripped
            current_section = ""  # reset section on new chapter
            continue
            
        # 2. Check for Section / Lesson (e.g. "1.1 The Nature of History" or "Lesson 2:")
        sec_match = SECTION_RE.match(stripped)
        if sec_match and len(stripped) < 100:
            current_section = stripped
            continue
            
    return current_chapter, current_section
```
