```python
import fitz
```

```python
def inspect(path, pages=range(100, 105)):
    doc = fitz.open(path)
    print(f"{path}: {len(doc)} pages")
    for i in pages:
        text = doc[i].get_text()
        print(f"\n--- Page {i+1} ({len(text)} chars) ---")
        print(text[:800])

inspect("../data/raw/logic.pdf")
```
